3.1 Complete LSA comparison
Legend: ✓ permitted · ✗ blocked · D default summary route only · — not applicable
LSA	Normal	Stub	Totally Stub	NSSA	Totally NSSA
Type 1 – Router	✓	✓	✓	✓	✓
Type 2 – Network	✓	✓	✓	✓	✓
Type 3 – Summary	✓	✓	D	✓	D
Type 4 – ASBR Summary	✓	✗	✗	✗	✗
Type 5 – External	✓	✗	✗	✗	✗
Type 7 – NSSA External	—	—	—	✓	✓

Important details:
- Type 1 and Type 2 describe topology within an area.
- Type 3 advertises networks between areas.
- Type 4 identifies how to reach an ASBR in another area.
- Type 5 advertises external routes.
- Type 7 advertises external routes originated inside an NSSA.
A Type 7 LSA is not flooded into Area 0 as Type 7. The NSSA ABR translates eligible Type 7 LSAs into Type 5 LSAs.

Memory trick:
- Stub: Block 4 and 5.
- Totally Stub: Block 3, 4 and 5.
- NSSA: Block incoming 5, but allow local 7.
- Totally NSSA: Block 3, 4 and 5 except default Type 3, but allow local 7.
