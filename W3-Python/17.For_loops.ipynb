Python For Loops

A for loop is used for iterating over a sequence (that is either a list, a tuple, a dictionary, a set, or a string).

This is less like the for keyword in other programming languages, and works more like an iterator method as found in other object-orientated programming languages.

With the for loop we can execute a set of statements, once for each item in a list, tuple, set etc.


```python
fruits = ["apple", "banana", "cherry"]
for x in fruits:
  print(x) 
```

```python
fruits = ["apple", "banana", "cherry"]
for x in fruits:
  print(fruits) 
```

```python
protocols = ["ospf", "bgp", "isis", "rip" , "eigrp"]

for i in protocols:
    print(i)
```

The for loop does not require an indexing variable to set beforehand.

The range() Function

To loop through a set of code a specified number of times, we can use the range() function,

The range() function returns a sequence of numbers, starting from 0 by default, and increments by 1 (by default), and ends at a specified number.

Example

Using the range() function:

```python
for i in range(6):
    print(i)
```

```python
for ipv6 in range(10):
    print(ipv6)
```

```python
for ipv6 in range(1,10):
    print(ipv6)
```

```python
for ipv6 in range(1,100,5):
    print(ipv6)
```

Note that range(6) is not the values of 0 to 6, but the values 0 to 5.


```python
for ipv4 in range(15):
    print(ipv4)
```

```python
for ipv4 in range(1,15,4):
    print(ipv4)
```


The range() function defaults to 0 as a starting value, however it is possible to specify the starting value by adding a parameter: range(2, 6), which means values from 2 to 6 (but not including 6):


Example

Using the start parameter:

```python
for i  in range(2,10):
    print(i)
```

```python
for ipv4 in range(15,100):
    print(ipv4)
```

The range() function defaults to increment the sequence by 1, however it is possible to specify the increment value by adding a third parameter: range(2, 30, 3):

```python
for i in range(2,30,5):
    print(i)
```

```python
for i in range(4,90,8):
    print(i)
```

```python
for ipv4 in range(15,100, 10):
    print(ipv4)
```

Else in For Loop

The else keyword in a for loop specifies a block of code to be executed when the loop is finished:

Example

Print all numbers from 0 to 5, and print a message when the loop has ended:

```python
for x in range(6):
    print(x)
else:
    print("test")
```

```python
i = 0
while (i<6):
    print(i)
    i+=1
else:
    print("test")
```

```python
for ipv4 in range(15,50):
    print(ipv4)
else:
    print("The numbers are printed")
```

```python
for ipv4 in range(15,500,100):
    print(ipv4)
else:
    print("The numbers are printed")
```

```python
for x in range(9):
    print(x)
else:
    print("test")
```

Note: The else block will NOT be executed if the loop is stopped by a break statement.

Example

Break the loop when x is 3, and see what happens with the else block:

```python
for x in range(20):
    if x==10:
        break
    print(x)
else:
    print("test")
```

```python
for x in range(20):
    print(x)
    if x==10:
        break
    #print(x)
else:
    print("test")
```

```python
for x in range(20):
    if x==10:
        continue
    print(x)
else:
    print("test")
```

Nested Loops

A nested loop is a loop inside a loop.

The "inner loop" will be executed one time for each iteration of the "outer loop":

Example

Print each adjective for every fruit:

```python
adj = ["red", "big", "tasty"]
fruits = ["apple", "banana", "cherry"]

for x in adj:
  for y in fruits:
    print(x, y)
```

```python
route_reflector = ["bgp", "Split horizon", "unicast", "multicast" , "route reflector client"]

split_horizon = ["bgp", "200","IBGP", "EBGP", "Weight", "Local Preference"]

for i in route_reflector:
    for j in split_horizon:
        print(i,j)
```

The pass Statement

for loops cannot be empty, but if you for some reason have a for loop with no content, put in the pass statement to avoid getting an error.

```python
for x in [0, 1, 2]:
  pass

# having an empty for loop like this, would raise an error without the pass statement
```
