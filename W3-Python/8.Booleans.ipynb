Evaluate Values and Variables
The bool() function allows you to evaluate any value, and give you True or False in return,

```python
print(bool("yaswanth"))


print(bool(51515))
```

```python
print(bool(0))          # False
print(bool(0.0))        # False
print(bool(""))         # False (empty string)
print(bool(None))       # False
print(bool(False))      # False
print(bool([]))         # False (empty list)
print(bool(()))         # False (empty tuple)
print(bool({}))         # False (empty dictionary)
print(bool(set()))      # False (empty set)
```

```python
print(bool("yaswanth"))  # True
print(bool(" "))         # True
print(bool(51515))       # True
print(bool(-10))         # True
print(bool(1))           # True
print(bool([1, 2]))      # True
print(bool({"a": 1}))    # True
```

```python
x="yaswantgh"
y=6566556

print(bool(x))

print(bool(y))
```

```python
print(bool("0"))
print(bool(0))
print(bool("False"))
print(bool(False))
print(bool([]))
print(bool([0]))
```
