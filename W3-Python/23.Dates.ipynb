Python Dates

A date in Python is not a data type of its own, but we can import a module named datetime to work with dates as date objects.


example:

Import the datetime module and display the current date:

```python
import datetime

x = datetime.datetime.now()

print(x)
```

```python
import datetime

dhcp = datetime.datetime.now()

print(dhcp)
```

```python
import datetime

psu = datetime.datetime.now()

print(psu)
```

Date Output

When we execute the code from the example above the result will be:

2025-11-29 18:40:40.336449

The date contains year, month, day, hour, minute, second, and microsecond.

The datetime module has many methods to return information about the date object.

Here are a few examples, you will learn more about them later in this chapter:

Example

Return the year and name of weekday:

```python
import datetime

x = datetime.datetime.now()

print(x.year)
print(x.strftime("%A"))
```

```python
import datetime


vlan = datetime.datetime.now()

print(vlan.year)
print(vlan.strftime("%A"))
print(vlan.strftime("%A"))
```

```python
import datetime


vlan = datetime.datetime.now()

print(vlan.year)
print(vlan.strftime("%A"))
print(vlan.strftime("%Y"))
print(vlan.strftime("%y"))
print(vlan.strftime("%m"))
print(vlan.strftime("%B"))
print(vlan.strftime("%b"))
print(vlan.strftime("%d"))
print(vlan.strftime("%a"))
print(vlan.strftime("%H"))
print(vlan.strftime("%I"))
print(vlan.strftime("%M"))
print(vlan.strftime("%S"))
print(vlan.strftime("%p"))
```

Creating Date Objects

To create a date, we can use the datetime() class (constructor) of the datetime module.

The datetime() class requires three parameters to create a date: year, month, day.

Example

Create a date object:

```python
import datetime

x = datetime.datetime(2020, 5, 17)

print(x)
```

```python
import datetime

isis = datetime.datetime(2026, 11, 26)

print(isis)
```

```python
import datetime

isis = datetime.datetime(2025, 10, 26)

print(isis)
```

The datetime() class also takes parameters for time and timezone (hour, minute, second, microsecond, tzone), but they are optional, and has a default value of 0, (None for timezone).

```python
import datetime

x = datetime.datetime(2018, 6, 1)

print(x.strftime("%B"))
```
