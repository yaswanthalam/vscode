Python - Unpack Tuples

Unpacking a Tuple

When we create a tuple, we normally assign values to it. This is called "packing" a tuple:


```python
fruits = ("apple", "banana", "cherry")

print(fruits)
```

```python
abc = ("yaswanth" ,"alam" , "alam" , 41,41, "85", "85" , "4151515", 0,5451,"and","jk")

print(abc)
```

```python
ISIS = ("level-1", "level-2" , "level-1-2" , "TLV" , "Area" , "Data base", "LSP", "LLH")

OSPF = ("area", "LSA", "LSDB", "LSR", "LSU")

print(OSPF)
```

```python
ISIS = ("level-1", "level-2" , "level-1-2" , "TLV" , "Area" , "Data base", "LSP", "LLH")

OSPF = ["area", "LSA", "LSDB", "LSR", "LSU"]

print(OSPF)
```

But, in Python, we are also allowed to extract the values back into variables. This is called "unpacking":

Example

Unpacking a tuple:

```python
fruits = ("apple", "banana", "cherry")

(green, yellow, red) = fruits

print(green)
print(yellow)
print(red)
```

```python
ISIS = ("level-1", "level-2" , "level-1-2")

(area, LSA, LSDB) = ISIS

print(area)
print(LSA)
print(LSDB)
```

Note: The number of variables must match the number of values in the tuple, if not, you must use an asterisk to collect the remaining values as a list.

```python
abc = ("alam" , "ab" , 545, "5451")

(alam, ab , xn , mn) = abc

print(alam)
print(ab)
print(xn)
print(mn)

print(alam, ab, xn, mn)
```

Using Asterisk*

If the number of variables is less than the number of values, you can add an * to the variable name and the values will be assigned to the variable as a list:

Example

Assign the rest of the values as a list called "red":

```python
fruits = ("apple", "banana", "cherry", "strawberry", "raspberry")

(green, yellow, *red) = fruits

print(green)
print(yellow)
print(red)
```

```python
ISIS = ("level-1", "level-2" , "level-1-2" , "TLV" , "Area" , "Data base", "LSP", "LLH")

(LSA, LSDB , *SPF) = ISIS

print(LSA)
print(LSDB)
print(SPF)
```

Using a While Loop

You can loop through the tuple items by using a while loop.

Use the len() function to determine the length of the tuple, then start at 0 and loop your way through the tuple items by referring to their indexes.

Remember to increase the index by 1 after each iteration.

Example

Print all items, using a while loop to go through all the index numbers:

```python
jm = ("asd" ,"asdfsd" ,"sfnbsdfj" ,"xfndf" ,"xffdnj",62165125, "545151")

i = 0

while i < len(jm):
    print(jm[i])
    i = i+1
```

```python
bc = ("asd" ,"asdfsd" ,"sfnbsdfj" ,"xfndf" ,"xffdnj",62165125, "545151")

i = 0

while i < len(bc):
    print(bc[i])
    i = i +1
```

Tuple Methods

Python has two built-in methods that you can use on tuples.

Method	Description

count()	Returns the number of times a specified value occurs in a tuple

index()	Searches the tuple for a specified value and returns the position of where it was found




```python
ISIS = ("level-1", "level-2" , "level-1-2" , "TLV" , "Area" , "Data base", "LSP", "LLH")

print(ISIS.index(2))
```

```python
ISIS = ("level-1", "level-2" , "level-1-2" , "TLV" , "Area" , "Data base", "LSP", "LLH")

print(ISIS.index[2])
```

```python
ISIS = ("level-1", "level-2" , "level-1-2" , "TLV" , "Area" , "Data base", "LSP", "LLH")

print(ISIS.index["level-1"])
```

```python
ISIS = ("level-1", "level-2" , "level-1-2" , "TLV" , "Area" , "Data base", "LSP", "LLH")

print(ISIS.index("level-1"))
```

```python
ISIS = ("level-1", "level-2", "level-1-2", "TLV",
        "Area", "Data base", "LSP", "LLH")

print(ISIS.index("level-1"))
print(ISIS.index("level-2"))
print(ISIS.index("level-1-2"))
print(ISIS.index("LSP"))
```

```python
ISIS = ("level-1", "level-2", "level-1-2", "TLV",
        "Area", "LSP", "LSP", "LLH")

print(ISIS.count("LSP"))
print(ISIS.count("level-1"))
print(ISIS.count("Area"))
print(ISIS.count("BGP"))
```

Count and Index are the 2 inbuilt functions we can use in tuples by default and they come with python  by default
