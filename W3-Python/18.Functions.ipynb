# Python Functions: Positional Arguments, Keyword Arguments & Mixing Both

## 1. Positional Arguments

A positional argument is an argument passed to a function based on its **position or order**.

Python assigns values to function parameters from left to right.

### Example

```python
def yas(name, age, job):
    print(name)
    print(age)
    print(job)

yas("Alam", 26, "Network Engineer")
```

**Output:**
```text
Alam
26
Network Engineer
```

### How it works

| Parameter | Position | Assigned value |
|---|---|---|
| name | 1st | Alam |
| age | 2nd | 26 |
| job | 3rd | Network Engineer |

**Remember:** Positional arguments depend on the order in which values are passed.

---

## 2. Keyword Arguments

A keyword argument is an argument passed using the parameter's name.

Python identifies which parameter receives the value using its name instead of its position.

### Example

```python
def yas(name, age, job):
    print(name)
    print(age)
    print(job)

yas(job="Network Engineer", name="Alam", age=26)
```

**Output:**
```text
Alam
26
Network Engineer
```

### How it works

Even though the arguments are passed in a different order, Python assigns the correct values because we explicitly specify parameter names.

**Remember:** Keyword arguments do not depend on the order in which they are passed.

---

## 3. Mixing Positional and Keyword Arguments

Python allows us to combine positional and keyword arguments in the same function call.

**Important rule: Positional arguments must always come before keyword arguments.**

### Example

```python
def yas(test, net, yak):
    print("I have an ", test,
          "betejs jxnfgjk", net,
          "oxford dictionary", yak)

yas("How are you ?", net=89,
    yak="ALAM Is the Enginner")
```

**Output:**
```text
I have an How are you ? betejs jxnfgjk 89 oxford dictionary ALAM Is the Enginner
```

### Parameter and argument mapping

| Parameter | Argument type | Assigned value |
|---|---|---|
| test | Positional | "How are you ?" |
| net | Keyword | 89 |
| yak | Keyword | "ALAM Is the Enginner" |

### Execution steps

1. Python reads the first positional argument: `"How are you ?"`.
2. It assigns this value to the first parameter, `test`.
3. Python reads `net=89` and assigns 89 to `net`.
4. Python reads `yak=...` and assigns the string to `yak`.
5. The function executes the print statement.

---

## 4. Why Must Positional Arguments Come First?

### Correct syntax

```python
def yas(test, net, yak):
    print(test, net, yak)

yas("Hello", net=89, yak="Engineer")
```

### Incorrect syntax

```python
yas(test="Hello", 89, "Engineer")
```

**Why does it fail?**

Python does not allow positional arguments after keyword arguments in a function call.

Correct order:

```python
function(positional_argument, keyword_argument)
```

---

## 5. Common Mistakes

### Mistake 1: Assigning the same parameter twice

```python
def yas(test, net, yak):
    print(test, net, yak)

yas("Hello", test="World", yak="Engineer")
```

**Error:**
```text
TypeError: yas() got multiple values for argument 'test'
```

**Reason:**

- `"Hello"` already assigns a value to `test`.
- `test="World"` attempts to assign another value to the same parameter.
- Python raises a TypeError.

### Mistake 2: Using an incorrect keyword name

```python
def yas(name, age, job):
    print(name, age, job)

yas("Alam", salary=90000, job="Engineer")
```

**Reason:** `salary` is not a parameter defined in the function.

---

## 6. Real-World Network Engineering Example

Consider a function that configures a BGP neighbor.

```python
def configure_bgp(neighbor, remote_as, description):

    print("Neighbor IP:", neighbor)
    print("Remote AS:", remote_as)
    print("Description:", description)


configure_bgp(
    "10.1.1.2",
    remote_as=65002,
    description="DC1-BGP-Peer"
)
```

**Output:**
```text
Neighbor IP: 10.1.1.2
Remote AS: 65002
Description: DC1-BGP-Peer
```

### Alternative: Using only keyword arguments

```python
configure_bgp(
    neighbor="10.1.1.2",
    remote_as=65002,
    description="DC1-BGP-Peer"
)
```

This approach is useful when functions have many parameters because it makes the code easier to read and maintain.

---

## 7. Comparison: Positional vs Keyword Arguments

| Feature | Positional arguments | Keyword arguments |
|---|---|---|
| Assignment | Based on position | Based on parameter name |
| Order | Important | Can be changed |
| Syntax | `yas("Alam", 26)` | `yas(name="Alam", age=26)` |
| Readability | Shorter | More descriptive |
| Mixing | Must come first | Must come after positional arguments |

---

## 8. Important Rules to Remember

1. Positional arguments are assigned from left to right.
2. Keyword arguments are assigned using parameter names.
3. Positional arguments must come before keyword arguments.
4. You cannot assign multiple values to the same parameter in one function call.
5. Keyword names must match the function's parameter names.
6. You can use only positional arguments, only keyword arguments, or mix both.

## 9. Quick Revision

```python
# Positional
yas("Alam", 26)

# Keyword
yas(name="Alam", age=26)

# Mixing
yas("Alam", age=26)
```

**Golden Rule:**

- Positional = Order matters.
- Keyword = Parameter name matters.
- Mixing = Positional first, keyword second.

Python Scope

Scope

A variable is only available from inside the region it is created. This is called scope.

Local Scope


A variable created inside a function belongs to the local scope of that function, and can only be used inside that function.

Example

A variable created inside a function is available inside that function:

````python
def myfunc():
  x = 300
  print(x)

myfunc()
````

````python
def alam():
    x=52156515
    print(x)

alam()
````

````python
def yas():
    y=5222
    z=960212
    x=5456215
    k=x+y-z
    print(k)
yas()
````


Global Scope

A variable created in the main body of the Python code is a global variable and belongs to the global scope.

Global variables are available from within any scope, global and local.

Example

A variable created outside of a function is global and can be used by anyone:


````python
x = 300

def myfunc():
  print(x)

myfunc()

print(x)
````

````python
y = 5000

def alam():
    print(y)

alam()
````

Naming Variables

If you operate with the same variable name inside and outside of a function, Python will treat them as two separate variables, one available in the global scope (outside the function) and one available in the local scope (inside the function):

Example

The function will print the local x, and then the code will print the global x:

````python
x = 300

def myfunc():
  x = 200
  print(x)

myfunc()

print(x)
````

````python
y = "sfgjdf"
def test():
    y = "shbdfjsf"
    print(y)
test()
print(y)
````


Global Keyword

If you need to create a global variable, but are stuck in the local scope, you can use the global keyword.

The global keyword makes the variable global.

Example


If you use the global keyword, the variable belongs to the global scope:


````python
def myfunc():
  global x
  x = 300

myfunc()

print(x)
````

````python
def you():
    global x
    x= "sdhbfjsfn"

you()
print(x)
````

Also, use the global keyword if you want to make a change to a global variable inside a function.

Example

To change the value of a global variable inside a function, refer to the variable by using the global keyword:

````python
a = 544652156

def rose():
    global a
    a = 100

rose()
print(a)
````


Nonlocal Keyword

The nonlocal keyword is used to work with variables inside nested functions.

The nonlocal keyword makes the variable belong to the outer function.

Example

If you use the nonlocal keyword, the variable will belong to the outer function:



````python
def myfunc1():
  x = "Jane"
  def myfunc2():
    nonlocal x
    x = "hello"
  myfunc2()
  return x

print(myfunc1())
````

````python
def test():
    x = ",moneky"
    def test1():
        nonlocal x
        x = "crocdi;e"
    test1()
    return x

print(test())
````

The LEGB Rule

Python follows the LEGB rule when looking up variable names, and searches for them in this order:

Local - Inside the current function

Enclosing - Inside enclosing functions (from inner to outer)

Global - At the top level of the module

Built-in - In Python's built-in namespace


Example

Understanding the LEGB rule:

````python
x = "global"

def outer():
  x = "enclosing"
  def inner():
    x = "local"
    print("Inner:", x)
  inner()
  print("Outer:", x)

outer()
print("Global:", x)
````
