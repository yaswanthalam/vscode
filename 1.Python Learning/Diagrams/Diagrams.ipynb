````python

from diagrams import Diagram
from diagrams.onprem.client import Users
from diagrams.onprem.compute import Server
from diagrams.onprem.database import PostgreSQL

with Diagram(
    "Simple Web Application",
    show=False,
    filename="web_application",
    outformat="png"
):
    Users("User") >> Server("Web Server") >> PostgreSQL("Database")

print("Diagram generated successfully!")
````

````python

import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(11, 4))

nodes = [
    (0.15, "User"),
    (0.50, "Web Server"),
    (0.85, "PostgreSQL")
]

for x, label in nodes:
    ax.text(
        x, 0.5, label,
        ha="center",
        va="center",
        fontsize=13,
        bbox=dict(
            boxstyle="round,pad=0.8",
            facecolor="lightblue",
            edgecolor="black"
        )
    )

for start, end in [(0.24, 0.41), (0.59, 0.76)]:
    ax.annotate(
        "",
        xy=(end, 0.5),
        xytext=(start, 0.5),
        arrowprops=dict(arrowstyle="->", lw=2)
    )

ax.set_xlim(0, 1)
ax.set_ylim(0, 1)
ax.axis("off")

plt.title("Simple Web Application Architecture")
plt.savefig("web_application.png", dpi=200)
plt.show()
````

````python

import plotly.graph_objects as go

fig = go.Figure()

nodes = [
    (0, "User"),
    (2, "Web Server"),
    (4, "PostgreSQL")
]

for x, name in nodes:
    fig.add_annotation(
        x=x, y=0,
        text=name,
        showarrow=False,
        bgcolor="lightblue",
        bordercolor="black",
        borderpad=15,
        font=dict(size=15)
    )

for x1, x2 in [(0.5, 1.5), (2.5, 3.5)]:
    fig.add_annotation(
        x=x2, y=0,
        ax=x1, ay=0,
        xref="x", yref="y",
        axref="x", ayref="y",
        text="",
        showarrow=True,
        arrowhead=3,
        arrowwidth=2
    )

fig.update_xaxes(range=[-1, 5], visible=False)
fig.update_yaxes(range=[-1, 1], visible=False)
fig.update_layout(
    title="Web Application Architecture",
    height=300
)

fig.show()
````

Yes! You can create architecture diagrams in Python without installing Graphviz. However, the `diagrams` library you are currently using requires Graphviz to render images.

## Which one should you use?

| Library      | Graphviz required? | Best for                                      |
| ------------ | ------------------ | --------------------------------------------- |
| `diagrams`   | Yes                | Cloud architecture with provider icons        |
| `matplotlib` | No                 | Simple static architecture diagrams           |
| Mermaid      | No                 | Network flows, architecture and documentation |
| Plotly       | No                 | Interactive diagrams                          |

My recommendation: For your network automation and AI-agent projects, use Mermaid for network topology diagrams and workflows. Use Matplotlib when you want to generate PNG diagrams entirely through Python.

One important distinction: none of these is a drop-in replacement for `diagrams`. You will need to rewrite the diagram definitions using the alternative library's syntax.
